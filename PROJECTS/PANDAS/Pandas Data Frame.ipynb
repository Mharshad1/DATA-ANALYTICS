```python
import pandas as pd
```

# Creating a DataFrame from a dictionary :

```python
data = {
     "Name" :  ["Anu","Rahul","Priya"],
    "Age"  : [21,22,21],
    "Mark"  : [85,90,88]
}
df = pd.DataFrame(data)
df
    
```

# Creating a DataFrame from a list :

```python
data = [
    ["Anu",21,85],
    ["Rahul",22,90],
    ["Priya",21,88]
]
df = pd.DataFrame(data ,columns =["Name","Age","Mark"])
df
```

# Creating a DataFrame with a custom index :

```python
data = {
     "Name" :  ["Anu","Rahul","Priya"],
     "Mark"  : [85,90,88]
}
df = pd.DataFrame(data , index = ["s1","s2","s3"])
df
```

# Selecting rows using loc :

```python
data = {
     "Name" :  ["Anu","Rahul","Priya"],
     "Mark"  : [85,90,88]
}
df = pd.DataFrame(data , index =["s1","s2","s3"])
df.loc["s2"]
```

# Adding a new column :

```python
df = pd.DataFrame({
     "Name" :  ["Anu","Rahul","Priya"],
     "Mark"  : [85,90,88]
})
df["Grade"] = ["A","A+","A"]
df
```

# Modifying a Existing column :

```python
df = pd.DataFrame({
     "Name" :  ["Anu","Rahul","Priya"],
     "Mark"  : [85,90,88]
})
df["Mark"] = [90,95,92]
df
```

# Adding a new row :

```python
df = pd.DataFrame({
     "Name" :  ["Anu","Rahul","Priya"],
     "Mark"  : [85,90,88]
})
df.loc[3] = ["Arun",80]
df
```

# Removing a column - drop() :

```python
df = df.drop("Mark" ,axis = 1)
df
```

# Removing a row - drop() :

```python
df = pd.DataFrame({
     "Name" :  ["Anu","Rahul","Priya"],
     "Mark"  : [85,90,88]
})
df.drop(1)
```

# Filtering rows :

```python
df = pd.DataFrame({
     "Name" :  ["Anu","Rahul","Priya","Arun"],
     "Mark"  : [85,90,45,75]
})
df[df["Mark"] > 80]
```

```python
df = df.sort_values("Mark")
df
```

# Descending order

# Checking for missing values - isna() :

```python
import pandas as pd
```

```python
import numpy as np
```

```python
df = pd.DataFrame({
     "Name" :  ["Anu","Rahul","Priya"],
     "Mark"  : [85,np.nan,90]
})
df.isna()
```

# Counting missing values :

```python
df.isna().sum()
```

# Filling missing values - fillna()

```python
df["Mark"] = df["Mark"].fillna(0)
df
```

# Removing rows with missing values : dropna() :

```python
df.dropna()
df
```
