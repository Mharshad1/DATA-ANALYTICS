```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "id" : [1,2,3,4],
    "Name" : ["Harshad","Monu","Fanu","Ayan"],
    "Roll_no" : [10,20,30,40]
})
df2 = pd.DataFrame({
    "id" : [1,2,3,5],
    "Marks" : [25,56,89,50],
    "Tag" : [11,21,31,41]
})
merged_df = pd.merge(df1,df2,on = "id")
merged_df
```

# Inner join

```python
pd.merge(df1,df2, on = "id",how = "inner")
```

# Left join

```python
pd.merge(df1,df2, on = "id",how = "left")
```

# Right join

```python
pd.merge(df1,df2, on = "id",how = "right")
```

# Outer join

```python
pd.merge(df1,df2, on = "id",how = "outer")
```

# Merge on different column names

```python
df1 = pd.DataFrame({
    "id" : [1,2,3,4],
    "Name" : ["Harshad","Monu","Fanu","Ayan"],
    "Roll_no" : [10,20,30,40]
})
df2 = pd.DataFrame({
    "student_id" : [1,2,3,5],
    "Marks" : [25,56,89,50],
    "Tag" : [11,21,31,41]
})
merged_df = pd.merge(df1,df2,left_on = "id",right_on = "student_id")
merged_df
```

# Concatenation

```python
pd.concat([df1,df2],axis = 0)
```

```python
pd.concat([df1,df2],axis = 1)
```

# Apply

```python
mark = pd.Series([30,50,45,70,90])
def grade(x):
    if x == 50 :
        return "Perfect match"
    else:
        return "Not match"
mark.apply(grade)
```

# Apply using Lambda Function

```python
mark = pd.Series([30,50,45,70,90])
grade = mark.apply(lambda x:x * 2)
grade
```

# Apply on dataframe columns

```python
mark = pd.DataFrame({
    "Physics" : [30,50,29,70,90],
    "Biology" : [20,40,60,80,100],
    "Chemistry" : [25,35,45,55,67]
})
print(mark.apply(lambda x:x.mean()))
```
