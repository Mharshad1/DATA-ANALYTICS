# Merging data using pandas :

# Basic merging

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "id" : [1,2,3],
    "Name" : ["Asha","Binu","Cathy"]
})
df2 = pd.DataFrame({
    "id" : [1,2,4],
    "Marks" : [85,90,75]
})
merged_df = pd.merge(df1,df2, on = "id")
merged_df
```

# Types of Joins :

# Inner Join (default) :

# only common rows

```python
pd.merge(df1,df2, on = "id", how = "inner")
```

# Left Join :

# All rows from leftdataframe

```python
pd.merge(df1,df2, on = "id" , how = "left")
```

# Right Join :

# All rows from rightdataframe

```python
pd.merge(df1,df2, on = "id" , how = "right")
```

# Outer Join :

# All rows from both

```python
pd.merge(df1,df2, on = "id" , how = "outer")
```

# Merge on Different Column Names :

```python
df1 = pd.DataFrame({
    "id" : [1,2,3],
    "Name" : ["Asha","Binu","Cathy"]
})
df2 = pd.DataFrame({
    "student_id" : [1,2,4],
    "Marks" : [85,90,75]
})
merged_df = pd.merge(df1,df2, left_on = "id", right_on = "student_id")
merged_df
```

# Concatenation :

```python
pd.concat([df1,df2],axis = 0) # row-wisw
```

```python
pd.concat([df1,df2],axis = 1) # column-wise
```

# Apply :

```python
marks = pd.Series([45,60,75,90])
def grade(x) :
    if x >= 50 :
        return "Pass"
    else:
        return "Fail"
marks.apply(grade)
```

# Apply using Lambda Function 

```python
marks = pd.Series([45,60,75,90])
result = marks.apply(lambda x:x + 5)
result
```
