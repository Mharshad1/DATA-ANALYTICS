```python
import pandas as pd
```

# Creating a Series :

# Series from a list

```python
numbers = [10,20,30,40]
num_series = pd.Series(numbers)
num_series
```

# Creating a series of list in strings

```python
s = pd.Series(["Apple","Orange","Banana","Kiwi"])
s
```

# Series with custom index

```python
marks = [85,90,75]
marks_series = pd.Series(marks,index = ["Rahul","Anu","John"])
marks_series
```

# Creating a series from a dictionary

```python
Students = {
    "Rahul" : 85,
    "Anu"  : 90,
    "John" : 75
}
student_series = pd.Series(Students)
student_series
```

# Sorting a Series :

# Ascending order

```python
s = pd.Series([25,10,45,30,15])
s.sort_values()
```

# Descending order

```python
s = pd.Series([25,10,45,30,15])
s.sort_values(ascending = False)
```

# Resetting the index after sorting

```python
s = pd.Series([25,10,45,30,15])
s.sort_values().reset_index(drop = True) 
                            
```

# Missing values in a series :

```python
import pandas as pd
import numpy as np
s = pd.Series([10,20,np.nan,40,np.nan])
s
```

# Checking for missing values : isna()

```python
s.isna()
```

# Removing missing values : dropna()

```python
s.dropna()
```

# Filling missing values : fillna()

```python
s = pd.Series([10,20,np.nan,40])
s.fillna(0)
```

# Adding two series together :

```python
s1 = pd.Series([10,20,30,40])
s2 = pd.Series([20,40,60,80])
s1 + s2
```

# Series VS a Plain list :

```python
numbers = [10,20,30,40]  # a plain list
numbers = pd.Series([10,20,30,40])   # series
print(numbers.mean())
print(numbers.max())
print(numbers.min())
print(numbers.sort_values())
```
