## <center> <h1 style="color:white; background-color:black; padding:8px;">Introduction</h1>
### COVID-19 is a respiratory disease caused by the SARS-CoV-2 coronavirus. It was first reported in Wuhan, China, in late December 2019, and the World Health Organization declared it a global pandemic in March 2020. This project uses a country-level COVID-19 dataset with 54,522 records covering 194 countries or regions, with one row for each country on each date, from 31 December 2019 to 6 November 2020. Each record contains the cumulative confirmed cases, deaths and recoveries, along with the new cases, deaths and recoveries reported that day. By the end of the period, the data shows about 49 million confirmed cases and over 1.2 million deaths worldwide. Because the dataset is large, covers many countries over time and contains real-world issues such as text-formatted dates, uneven coverage and negative correction values, it is well suited to a data manipulation project on cleaning, filtering, grouping and transforming data.

## <center> <h1 style="color:white; background-color:black; padding:8px;">About The Dataset</h1>
### The dataset used in this project, *COVID_Data_Basic.csv*, is a country-level COVID-19 time series with 54,522 records for 194 countries and regions, with one row for each country on each date. It covers 31 December 2019 to 6 November 2020, which includes the first wave of the pandemic and the start of the second. It has nine columns: a row index, the country name, the date, three cumulative totals (confirmed cases, deaths and recoveries) and three daily figures (new cases, new deaths and new recoveries). By the end of the period, the data shows roughly 49 million confirmed cases and 1.24 million deaths worldwide. There are no missing values, but the data needs cleaning because the dates are stored as text, some country names are inconsistent, coverage is uneven across countries and a few daily values are negative. This makes it a realistic dataset for practising data manipulation tasks such as cleaning, filtering, grouping and creating calculated columns.

## <center> <h1 style="color:white; background-color:black; padding:8px;">Conclusion</h1>
### This project cleaned and analysed a country-level COVID-19 dataset of 54,522 records covering 194 countries and regions, from 31 December 2019 to 6 November 2020. After dropping the unnecessary index column, the data had no missing values and no duplicate records. The analysis shows that cases were concentrated in a few countries. The US had the most confirmed cases (9,733,816) and the most deaths (236,073), followed by India (8,462,080 cases) and Brazil (5,631,181 cases), with Russia and France next. India had the most recoveries (7,819,886), followed by Brazil. The highest single-day increase in cases was in the US (126,480), and Peru had the highest single-day death count (4,143), which may reflect a one-off revision of earlier figures. Across all countries, the daily new cases add up to about 49.3 million confirmed cases by 6 November 2020. Overall, the pandemic hit a small number of large countries hardest, and the data was still showing rising cases at the end of the period.
