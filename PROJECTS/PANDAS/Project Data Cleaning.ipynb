# <center> Introduction

This project focuses on cleaning the **Netflix Titles dataset** (`NetFlix-selected-columns.csv`), which contains metadata for 7,787 movies and TV shows available on Netflix. The dataset includes 10 attributes describing each title — such as `show_id`, `type`, `title`, `director`, `cast`, `country`, `date_added`, `release_year`, `rating`, and `duration` — offering a rich mix of categorical, textual, and numerical data suitable for a real-world data cleaning exercise.

Despite being a well-known public dataset, it is far from clean in its raw form. It suffers from common real-world data quality issues, including missing values in key columns (`director`, `cast`, `country`, `date_added`, and `rating`), inconsistent date formatting in `date_added`, an ambiguous `duration` field that represents different units depending on content type (minutes for movies vs. seasons for TV shows), and multi-valued fields (`cast`, `country`) packed into single cells.

The goal of this project is to systematically identify, assess, and resolve these inconsistencies — handling missing data appropriately, standardizing formats, and restructuring ambiguous or compound fields — in order to produce a clean, analysis-ready version of the dataset. This process reflects the essential first step of any real-world data analysis workflow, where raw data must be transformed into a reliable and consistent form before meaningful insights can be drawn.

```python
import pandas as pd
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
df["director"].fillna("unknown" , inplace = True)
```

```python
df["cast"].fillna(df["cast"].mode()[0] , inplace = True)
```

```python
df["country"].fillna(df["country"].mode()[0] , inplace = True)
```

```python
df["date_added"].fillna(df["date_added"].mode()[0] , inplace = True)
```

```python
df["rating"].fillna(df["rating"].mode()[0] , inplace = True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.to_csv("New_Netflix.csv", index = False)
```

# <center> Conclusion

Before diving into cleaning, I explored the dataset using `shape`, `size`, `info()`, `describe()`, `head()`, and `tail()` to understand its structure, size, and data types. This exploration showed that the dataset is mostly categorical in nature, with `release_year` being the main numerical column worth statistical summary. The `info()` step also revealed missing values in six columns — `director`, `cast`, `country`, `date_added`, and `rating` — totaling 3,631 null entries. I then checked for duplicate rows using `duplicated().sum()`, which confirmed that the dataset had **no duplicates**, meaning every record is unique.

To handle the missing values, I filled `director` with "unknown" since a large portion of its values were missing and no reliable substitute existed, while `cast`, `country`, `date_added`, and `rating` were filled using their respective mode (most frequent value), as these columns had comparatively fewer missing entries. After this treatment, `isnull().sum()` confirmed that the dataset now has zero missing values across all columns.

Overall, the dataset started out unique but incomplete, and after handling the null values through appropriate imputation techniques, it is now both clean and duplicate-free — making it ready for further analysis or visualization.
