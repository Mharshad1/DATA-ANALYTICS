# Create and write to file

```python
with open("example.txt","w") as file:
    file.write("File is created successfully")
```

# Read file

```python
with open("example.txt","r") as file:
    print(file.read())
```

# Overwrite file

```python
with open("example.txt","w") as file:
    file.write("Hello, it is created successfully")
```

# Append new line

```python
with open("example.txt","a") as file:
    file.write("\nAdding a new line.")
```

# Read full content again

```python
with open("example.txt","r") as file:
    print(file.read())
```

# Read line by line

```python
with open("example.txt","r") as file:
    for line in file:
        print(line.strip())
```

# Read only one line

```python
with open("example.txt","r") as file:
    print(file.readline())
```

# Working with file pointer

```python
with open("example.txt","r") as file:
    print(file.read(5))
    file.seek(0)
    print(file.read())
    
```

# Tell current position

```python
with open("example.txt","r") as file:
    print(file.tell())
    file.read(10)
    print(file.tell())
    
```

# Read lines as list

with open("example.txt","r") as file:
    print(file.readlines())

# Writing multiple lines

```python
a = ["Hai,My name is Harshad\n" ,"How are you?\n" , "How old are you?\n"]
with open("example.txt","w") as file:
    file.writelines(a)
    
```

# File copy

```python
with open("example.txt","r") as src:
    a = src.read()
with open("copy example.txt","w") as dest:  
    dest.write(a)
    
```

# Rename file

```python
import os
os.rename("example.txt","new_example.txt")
    
```

# Delete file

```python
import os
os.remove("new_example.txt")
    
```

# Working with CSV file

```python
import csv
with open("data.csv","w", newline = "") as file:
    a = csv.writer(file)
    a.writerow(["Name","Age"])
    a.writerow(["Harshad",23])
```

# Working with json file

```python
import json
data = {"Name":"Harshad","Age":23}
with open("data.json","w") as file: 
    json.dump(data ,file)
with open("data.json","r") as file: 
    json.load(file)    
```
