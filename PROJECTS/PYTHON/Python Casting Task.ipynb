```python
int_num = 10
float_num = 25.55
str_num = "29"
bool_num = True
complex_num = 10 + 2j

converted_int = float(int_num)
print("Original:",int_num,type(int_num))
print("Converted:",converted_int,type(converted_int))

converted_float = int(float_num)
print("\nOriginal:",float_num,type(float_num))
print("Converted:",converted_float,type(converted_float))

converted_str = int(str_num)
print("\nOriginal:",str_num,type(str_num))
print("Converted:",converted_str,type(converted_str))

converted_bool = int(bool_num)
print("\nOriginal:",bool_num,type(bool_num))
print("Converted:",converted_bool,type(converted_bool))

converted_complex = str(complex_num)
print("\nOriginal:",complex_num,type(complex_num))
print("Converted:",converted_complex,type(converted_complex))
```
