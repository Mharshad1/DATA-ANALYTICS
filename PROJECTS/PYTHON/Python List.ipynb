# Ascending

```python
Thislist = [10 ,50, 32 , 12, 47, 63, 90]
Thislist.sort()
print(Thislist)
```

# Descending

```python
Thislist = [10 ,50, 32 , 12, 47, 63, 90]
Thislist.sort(reverse = True)
print(Thislist)
```

# Reverse

```python
Thislist = [10 ,50, 32 , 12, 47, 63, 90]
Thislist.reverse()
print(Thislist)
```

```python
Thislist = [10 ,50, 32 , 12, 47, 63, 90]
Thislist.sort(reverse = False)
print(Thislist)
```
