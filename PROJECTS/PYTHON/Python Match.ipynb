# Match

```python
Department = input("Enter the department:(CS/EC/IT/ME)").upper()
match Department:
    case "CS":
        print("Computer Science")
    case "EC":
        print("Electronics")
    case "IT":
        print("Information Technology")
    case "ME":
        print("Mechanical Engineering")
    case _:
        print("Invalid code")
```

```python
Month = int(input("Enter the month:"))
match Month:
    case 1:
        print("Jan")
    case 2:
        print("Feb")
    case 3:
        print("Mar")
    case 4:
        print("Apr")
    case 5:
        print("May")
    case 6:
        print("Jun")
    case 7:
        print("Jul")
    case 8:
        print("Aug")
    case 9:
        print("Sep")
    case 10:
        print("Oct")
    case 11:
        print("Nov")
    case 12:
        print("Dec")
```

```python
num1 = int(input("Enter the num1:"))
num2 = int(input("Enter the num2:"))
Op = input("Enter the operator:(+ , - , * , / )")
match Op:
    case "+":
        print("Result is:",num1 + num2)
    case "-":
        print("Result is:",num1 - num2)
    case "*":
        print("Result is:",num1 * num2)
    case "/":
        if num2 != 0 :
            print("Result is:",num1 / num2)
        else :
            print("Division is not possible")
    case _:
        print("Invalid operator")
```

```python
Grade = input("Enter the grade:(A/B/C/D/E/F)").upper()
match Grade:
    case "A":
        print("Excellent")
    case "B":
        print("Good")
    case "C":
        print("Average")
    case "D":
        print("Pass")
    case "F":
        print("Fail")
    case _:
        print("Invalid")
```

```python
Month = int(input("Enter the month:"))
match Month:
    case 6 | 7 | 8 :
        print("Monsoon")
    case 9 | 10 | 11 :
        print("Spring")
    case 12 | 1 | 2 :
        print("Winter")
    case 3 | 4 | 5 :
        print("Summer")
    case _ :
        print("Invalid Month")
```

```python
Month = int(input("Enter the month:"))
match Month:
    case 1 | 3 | 5 | 7 | 8 | 10 | 12 :
        print("31 Days")
    case 2 :
        print("28 Days")
    case 4 | 6 | 9 | 11 :
        print("30 Days")
    case _ :
        print("Invalid Month")
```
