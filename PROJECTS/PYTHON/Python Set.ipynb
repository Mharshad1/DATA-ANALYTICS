# set

```python
Football = {"Harshad","Shan","Layan","Monu","Fanu"}
Cricket = {"Layan","Aiden","Harshad","Monu","Hadi"}
c = Football.union(Cricket)
print(c)
```

```python
Football = {"Harshad","Shan","Layan","Monu","Fanu"}
Cricket = {"Layan","Aiden","Harshad","Monu","Hadi"}
c = Football.symmetric_difference(Cricket)
print(c)
```

```python
Football = {"Harshad","Shan","Layan","Monu","Fanu"}
Cricket = {"Layan","Aiden","Harshad","Monu","Hadi"}
c = Football.intersection(Cricket)
print(c)
```

```python
Football = {"Harshad","Shan","Layan","Monu","Fanu"}
Cricket = {"Layan","Aiden","Harshad","Monu","Hadi"}
c = Football.difference(Cricket)
print(c)
```

# Find missing values using set

```python
a = {1,3,5,7,9,10,14,18,20}
b = range(1,21)
c = set(b)
d = c.difference(a)
print(d)
```

# Find common elements

```python
a = str(input("Enter the first string:"))
b = str(input("Enter the second string:"))
c = set(a) & set(b)
print("The common element 'a' and 'b' is :",c)
```
