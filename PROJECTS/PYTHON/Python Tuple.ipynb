# Tuple

```python
Mytuple = ("Green","yellow","blue","pink","red")
print(len(Mytuple))
```

```python
Mytuple = ("Green","yellow","blue","pink","red")
print(Mytuple[0])
print(Mytuple[-1])      
```

```python
Mytuple = (12,45,8,67,29,10,54)
for x in Mytuple:
    print(x)
```

```python
Mytuple = (12,45,8,67,29,10,54)
print(max(Mytuple))
print(min(Mytuple))
```

```python
Mytuple = (12,45,8,67,29,10,54)
print(87 in Mytuple)
```

```python
Mytuple = (12,45,8,67,29,10,54)
print(sum(Mytuple))
```

```python
Mytuple = (12,45,8,67,29,10,54)
a = list(Mytuple)
a.reverse()
Mytuple = tuple(a)
print(Mytuple)
```

```python
Mytuple = (12,45,8,67,29,10,54)
print(Mytuple[::-1])
```

```python
squares = []
for i in range (1,11):
    squares.append(i**2)
square_tuple = tuple(squares)
print(square_tuple)
```
