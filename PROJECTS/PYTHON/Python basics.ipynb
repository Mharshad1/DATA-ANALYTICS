# Match

```python
Department = input("Enter the department:(CS/EC/IT/ME)").upper()
match Department:
    case "CS":
        print("Computer Science")
    case "EC":
        print("Electronics")
    case "IT":
        print("Information Technology")
    case "ME":
        print("Mechanical Engineering")
    case _:
        print("Invalid code")
```

```python
Month = int(input("Enter the month:"))
match Month:
    case 1:
        print("Jan")
    case 2:
        print("Feb")
    case 3:
        print("Mar")
    case 4:
        print("Apr")
    case 5:
        print("May")
    case 6:
        print("Jun")
    case 7:
        print("Jul")
    case 8:
        print("Aug")
    case 9:
        print("Sep")
    case 10:
        print("Oct")
    case 11:
        print("Nov")
    case 12:
        print("Dec")
```

```python
num1 = int(input("Enter the num1:"))
num2 = int(input("Enter the num2:"))
Op = input("Enter the operator:(+ , - , * , / )")
match Op:
    case "+":
        print("Result is:",num1 + num2)
    case "-":
        print("Result is:",num1 - num2)
    case "*":
        print("Result is:",num1 * num2)
    case "/":
        if num2 != 0 :
            print("Result is:",num1 / num2)
        else :
            print("Division is not possible")
    case _:
        print("Invalid operator")
```
