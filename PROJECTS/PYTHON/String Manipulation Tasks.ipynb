# String Slicing

```python
char = str(input("Enter the character :"))
print(char[:5])
```

```python
char = str(input("Enter the character :"))
print(char[-4:])
```

```python
char = str(input("Enter the character :"))
print(char[2:7])
    
```

```python
char = str(input("Enter the character :"))
print(char[::2])
```
